```python
country_files = {
    'usa': 'USvideos',
    'ca':  'CAvideos',
    'de':  'DEvideos',
    'fr':  'FRvideos',
    'gb':  'GBvideos',
    'in':  'INvideos',
    'jp':  'JPvideos',
    'kr':  'KRvideos',
    'mx':  'MXvideos',
    'ru':  'RUvideos',
}

dfs = {}
for country, filename in country_files.items():
    df = pd.read_csv(os.path.join(os.getcwd(), 'YT file', f'{filename}.csv'),
                      encoding='utf-8', encoding_errors='replace')
    df['country'] = country
    dfs[country] = df
```

```python
## 5. Add a like_ratio column storing the ratio between the number of likes and of dislikes
```

```python
def perc(row):
    if row['dislikes'] == 0:
        return None
    return row['likes'] / row['dislikes']

full2['like_ratio'] = full2.apply(perc, axis=1)
full2.head(2)
```
